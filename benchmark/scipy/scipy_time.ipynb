```python
import numpy as np
from scipy.spatial import KDTree
from scipy.io import FortranFile as FF

file = '../examples/points.npy'

# Load the points from the numpy file
points = np.load(file)
n = points.shape[0]
```

```python
tree = KDTree(points)
```

```python
import time
import timeit
```

```python
target = [50., 50., 50.]

rad = 30
def query():
    return tree.query_ball_point(target, rad)

ntrials = 10
total_time = timeit.timeit(query, number=ntrials)

print(f"KDTree query time: {total_time / ntrials:.6e} seconds per query")
```

```python
target = [50., 50., 50.]
k = 100
def query():
    return tree.query(target, k=k)

ntrials = 10000
total_time = timeit.timeit(query, number=ntrials)

print(f"KDTree query time: {total_time / ntrials:.6e} seconds per query")
```
